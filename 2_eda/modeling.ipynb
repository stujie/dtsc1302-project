### Research Question

Which socioeconomic characteristics of a census tract (ex. income, poverty, housing costs, education, race, etc.) are associated with having a data center? Do the correlations change based on facility size and state?

### The Dataset

`clean.csv`: 10599 rows & 32 columns


### Set Up

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r'clean.csv')
df
```

### Split Data: Training & Test

```python
numeric_cols = ["median_hh_income", "median_gross_rent", "median_home_value",
                "poverty_rate", "pct_bachelors_plus", "pct_owner_occupied",
                "pct_no_vehicle", "pct_rent_burdened"]
categorical_cols = ["state"]

# Features and the target
features = categorical_cols + numeric_cols
target = "has_data_center"

train, test = train_test_split(df, test_size=0.2, random_state=42, stratify=df[target])

# Fill missing numeric values with the TRAINING median (so the test set stays unseen)
medians = train[numeric_cols].median()
train = train.copy(); test = test.copy()
train[numeric_cols] = train[numeric_cols].fillna(medians)
test[numeric_cols] = test[numeric_cols].fillna(medians)

print("Training rows:", len(train), "| Test rows:", len(test))
print("Missing values left:", train[features].isna().sum().sum() + test[features].isna().sum().sum())
print("\nMedians used to fill gaps:\n", medians[["median_gross_rent", "median_home_value", "median_hh_income"]])
print("\nData center tracts in training:", train[target].sum(), "| in test:", test[target].sum())
print(train[target].value_counts().sort_index())
```
